```python
import threading
import multiprocessing
import queue
import time


# ---------------- THREADING ----------------

def thread_producer(q):
    for i in range(1, 6):
        q.put(i)
        print("Thread Producer produced:", i)
        time.sleep(0.2)

    q.put(None)


def thread_consumer(q):
    while True:
        item = q.get()

        if item is None:
            break

        print("Thread Consumer consumed:", item)
        q.task_done()


def run_threading():
    print("\n--- THREADING ---")

    q = queue.Queue()

    producer = threading.Thread(target=thread_producer, args=(q,))
    consumer = threading.Thread(target=thread_consumer, args=(q,))

    producer.start()
    consumer.start()

    producer.join()
    consumer.join()


# ---------------- MULTIPROCESSING ----------------

def process_producer(q):
    for i in range(1, 6):
        q.put(i)
        print("Process Producer produced:", i)

    q.put(None)


def process_consumer(q):
    while True:
        item = q.get()

        if item is None:
            break

        print("Process Consumer consumed:", item)


def run_multiprocessing():
    print("\n--- MULTIPROCESSING ---")

    q = multiprocessing.Queue()

    producer = multiprocessing.Process(
        target=process_producer, args=(q,)
    )

    consumer = multiprocessing.Process(
        target=process_consumer, args=(q,)
    )

    producer.start()
    consumer.start()

    producer.join()
    consumer.join()


if __name__ == "__main__":
    run_threading()
    run_multiprocessing()

    print("\nProducer-Consumer execution completed successfully.")
```
